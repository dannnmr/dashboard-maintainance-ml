```python
# ============================================
# 0) Imports, config y utilidades
# ============================================
import os, json
from pathlib import Path
from typing import List
from collections import Counter, deque

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from deltalake import DeltaTable
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    classification_report, confusion_matrix, ConfusionMatrixDisplay,
    roc_auc_score, average_precision_score, precision_recall_curve, roc_curve
)
from sklearn.ensemble import IsolationForest
from sklearn.tree import DecisionTreeClassifier

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
import joblib

# Reproducibility
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)
plt.rcParams["figure.figsize"] = (6, 4)

# Window setup
LOOKBACK = 24          # number of time steps per sequence (e.g., 24 hours)
HORIZON_SHIFT = 12     # 0 for "now", 6/12/24 to predict hours ahead

# Paths
def find_repo_root() -> Path:
    p = Path.cwd().resolve()
    for cand in (p, *p.parents):
        if (cand / "data").exists():
            return cand
    return p

BASE_DIR = find_repo_root()
RUTA_GOLD_COMPLETE = BASE_DIR / "data" / "capa_gold" / "features_transformador" / "features_complete"

# Artifacts and results folders
ARTIFACTS_DIR = BASE_DIR / "data" / "capa_gold" / "artifacts_anomalia"
PLOTS_DIR = ARTIFACTS_DIR / "plots"
RESULTS_DIR = BASE_DIR / "data" / "capa_gold" / "artifacts_anomalia_results"
for d in (ARTIFACTS_DIR, PLOTS_DIR, RESULTS_DIR):
    d.mkdir(parents=True, exist_ok=True)

# Columns to exclude (labels / text / future-derived)
LEAK_OR_TEXT_COLS = [
    "estado_futuro","falla_30d","rul_dias","severidad_futura",
    "dias_proximo_evento","proximidad_evento","riesgo_acumulativo",
    "estado_operacional","nivel_severidad","variables_anomalas","descripcion_anomalia"
]
TARGET = "estado_futuro"

# --------------------------------------------
# Utils: plots, thresholds, smoothing, scaling
# --------------------------------------------
def save_show(figpath: Path):
    """Save current Matplotlib figure then show it."""
    plt.tight_layout()
    plt.savefig(figpath, dpi=160)
    plt.show()

def plot_roc_pr(y_true: np.ndarray, scores: np.ndarray, title_prefix: str, out_prefix: str):
    """Plot ROC and PR curves and save them."""
    fpr, tpr, _ = roc_curve(y_true, scores)
    prec, rec, _ = precision_recall_curve(y_true, scores)
    roc_auc = roc_auc_score(y_true, scores)
    pr_auc  = average_precision_score(y_true, scores)

    plt.figure()
    plt.plot(fpr, tpr); plt.plot([0,1],[0,1],'--')
    plt.title(f"{title_prefix} ROC (AUC={roc_auc:.3f})")
    plt.xlabel("FPR"); plt.ylabel("TPR")
    save_show(PLOTS_DIR / f"{out_prefix}_roc.png")

    plt.figure()
    plt.plot(rec, prec)
    plt.title(f"{title_prefix} Precision-Recall (AP={pr_auc:.3f})")
    plt.xlabel("Recall"); plt.ylabel("Precision")
    save_show(PLOTS_DIR / f"{out_prefix}_pr.png")

def eval_with_thr(scores: np.ndarray, y_true: np.ndarray, thr: float, title: str, out_prefix: str, cmap="Blues"):
    """Evaluate a binary threshold and plot confusion matrix."""
    yhat = (scores > thr).astype(int)  # 1 = anomalous
    print(f"\n{title} (thr={thr:.6f})")
    print(classification_report(y_true, yhat, target_names=["NORMAL","NO-NORMAL"], digits=4, zero_division=0))
    cm = confusion_matrix(y_true, yhat, labels=[0,1])
    ConfusionMatrixDisplay(cm, display_labels=["NORMAL","NO-NORMAL"]).plot(cmap=cmap)
    plt.title(title)
    save_show(PLOTS_DIR / f"{out_prefix}_cm.png")
    return yhat

def best_thr_fbeta(scores: np.ndarray, y_true: np.ndarray, beta: float = 2.0, grid: int = 400):
    """Find threshold that maximizes F_beta for positive class (NO-NORMAL)."""
    lo, hi = scores.min(), scores.max()
    ths = np.linspace(lo, hi, grid)
    best = (0.0, None, (0,0,0))  # (fbeta, thr, (precision, recall, f1))
    b2 = beta * beta
    for th in ths:
        yhat = (scores > th).astype(int)
        tp = ((y_true==1) & (yhat==1)).sum()
        fp = ((y_true==0) & (yhat==1)).sum()
        fn = ((y_true==1) & (yhat==0)).sum()
        prec = tp / (tp + fp) if (tp+fp)>0 else 0.0
        rec  = tp / (tp + fn) if (tp+fn)>0 else 0.0
        denom = b2*prec + rec
        fbeta = ((1+b2)*prec*rec/denom) if denom>0 else 0.0
        if fbeta > best[0]:
            f1 = (2*prec*rec)/(prec+rec) if (prec+rec)>0 else 0.0
            best = (fbeta, th, (prec, rec, f1))
    return best

def smooth_alerts(binary_seq: np.ndarray, k: int = 3, m: int = 5) -> np.ndarray:
    """Return 1 if at least k anomalies appear within a sliding window of length m."""
    out = np.zeros_like(binary_seq)
    win = deque(maxlen=m)
    count = 0
    for i, v in enumerate(binary_seq):
        if len(win) == m:
            count -= win[0]
        win.append(v)
        count += v
        out[i] = 1 if count >= k else 0
    return out

def minmax_transform(x: np.ndarray, lo: float, hi: float):
    """Min-max scaling with guard against zero range."""
    eps = 1e-12
    return (x - lo) / max(hi - lo, eps)

# ============================================
# 1) Load Gold and basic prep
# ============================================
dt = DeltaTable(str(RUTA_GOLD_COMPLETE))
df = dt.to_pandas()
df["timestamp"] = pd.to_datetime(df["timestamp"], utc=True, errors="coerce")
df = df.sort_values("timestamp").set_index("timestamp")

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
X_cols = [c for c in numeric_cols if c not in LEAK_OR_TEXT_COLS]

# Drop constant columns and those with >30% NaN (for column selection only)
const_cols = [c for c in X_cols if df[c].nunique() <= 1]
if const_cols:
    print("Drop constantes:", const_cols)
    X_cols = [c for c in X_cols if c not in const_cols]

nan_ratio = df[X_cols].isna().mean()
high_nan = nan_ratio[nan_ratio > 0.30].index.tolist()
if high_nan:
    print("Drop >30% NaN:", high_nan)
    X_cols = [c for c in X_cols if c not in high_nan]

X_full = df[X_cols].replace([np.inf, -np.inf], np.nan)
y_raw = df[TARGET].astype(str)

# Label encoding (fit on all labels to keep mapping intact)
le = LabelEncoder()
y_enc = le.fit_transform(y_raw)
print("Clases:", dict(zip(le.classes_, range(len(le.classes_)))))

# ============================================
# 2) Temporal split (80/20) without leakage
# ============================================
n = len(X_full); cut = int(n*0.80)
train_index = X_full.index[:cut]
valid_index = X_full.index[cut:]

# Impute NaN using TRAIN medians (no leakage)
train_medians = X_full.loc[train_index].median(numeric_only=True)
X_train_df = X_full.loc[train_index].fillna(train_medians)
X_valid_df = X_full.loc[valid_index].fillna(train_medians)

y_train_vec = le.transform(df.loc[train_index, TARGET].astype(str))
y_valid_vec = le.transform(df.loc[valid_index, TARGET].astype(str))

# NORMAL class id (fallback to majority if missing)
try:
    normal_id = int(np.where(le.classes_ == "NORMAL")[0][0])
except Exception:
    normal_id = Counter(y_train_vec).most_common(1)[0][0]

# Binary labels for validation (1 = NO-NORMAL)
y_valid_bin = (y_valid_vec != normal_id).astype(int)
print("\n== SPLIT ==")
print("Train:", X_train_df.shape, " Valid:", X_valid_df.shape)
print("Train range:", train_index.min(), "→", train_index.max())
print("Valid range:", valid_index.min(), "→", valid_index.max())
print("Distribución VALID (bin):", Counter(y_valid_bin))

# ============================================
# 3) Scaling (fit on NORMAL only) and sequences
# ============================================
mask_tr_normal = (y_train_vec == normal_id)

scaler_ae = StandardScaler()
X_train_sc = pd.DataFrame(
    scaler_ae.fit_transform(X_train_df[mask_tr_normal]),
    index=X_train_df.index[mask_tr_normal],
    columns=X_cols
)
X_valid_sc = pd.DataFrame(
    scaler_ae.transform(X_valid_df),
    index=X_valid_df.index,
    columns=X_cols
)

def make_sequences(X: pd.DataFrame, lookback: int):
    """Build rolling sequences and return (array, index_of_last_timestamps)."""
    Xv = X.values; N, F = Xv.shape
    seqs, idx = [], []
    for t in range(lookback-1, N-HORIZON_SHIFT):
        seqs.append(Xv[t-(lookback-1):t+1])
        idx.append(X.index[t])  # timestamp at window end
    return np.array(seqs, dtype=np.float32), pd.Index(idx, name="timestamp")

# AE sequences: train on NORMAL only
Xtr_seq, tr_idx = make_sequences(X_train_sc, LOOKBACK)
Xva_seq, va_idx = make_sequences(X_valid_sc, LOOKBACK)

# Align validation labels to sequences and horizon
yva_full = (y_valid_vec != normal_id).astype(int)  # binary
yva_bin_aligned = yva_full[LOOKBACK-1:len(yva_full)-HORIZON_SHIFT]

print("\n== SEQ ==")
print("Xtr_seq:", Xtr_seq.shape, "| Xva_seq:", Xva_seq.shape)
print("y_valid_bin aligned:", yva_bin_aligned.shape)

# ============================================
# 4) LSTM Autoencoder (train on NORMAL)
# ============================================
def build_lstm_ae(steps, feats):
    inp = layers.Input(shape=(steps, feats))
    x = layers.Masking()(inp)
    x = layers.LSTM(128, return_sequences=True)(x)
    x = layers.Dropout(0.2)(x)
    x = layers.LSTM(64)(x)
    z = layers.Dense(32, activation="relu")(x)
    x = layers.RepeatVector(steps)(z)
    x = layers.LSTM(64, return_sequences=True)(x)
    x = layers.LSTM(128, return_sequences=True)(x)
    out = layers.TimeDistributed(layers.Dense(feats))(x)
    m = models.Model(inp, out)
    m.compile(optimizer=tf.keras.optimizers.Adam(1e-3), loss="mse")
    return m

ae = build_lstm_ae(LOOKBACK, len(X_cols))
es  = callbacks.EarlyStopping(monitor="val_loss", patience=6, restore_best_weights=True)
rlr = callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-5)

hist = ae.fit(
    Xtr_seq, Xtr_seq,
    validation_data=(Xva_seq, Xva_seq),
    epochs=60, batch_size=128,
    callbacks=[es, rlr], verbose=1
)

# Training curves
plt.figure()
plt.plot(hist.history["loss"], label="Train loss")
plt.plot(hist.history["val_loss"], label="Valid loss")
plt.title("AE-LSTM Training vs Validation Loss")
plt.xlabel("Epoch"); plt.ylabel("MSE"); plt.legend()
save_show(PLOTS_DIR / "ae_training_loss.png")

# Reconstruction error as anomaly score
def recon_error(model, Xseq):
    rec = model.predict(Xseq, verbose=0)
    return np.mean((Xseq - rec)**2, axis=(1,2))

err_tr = recon_error(ae, Xtr_seq)  # for normalization & p95 threshold
err_va = recon_error(ae, Xva_seq)

# Threshold-free metrics (validation)
ae_roc_auc = roc_auc_score(yva_bin_aligned, err_va)
ae_pr_auc  = average_precision_score(yva_bin_aligned, err_va)
print(f"\n[AE-LSTM] ROC-AUC={ae_roc_auc:.3f} | PR-AUC={ae_pr_auc:.3f}")
plot_roc_pr(yva_bin_aligned, err_va, "AE-LSTM", "ae")

# Thresholds
thr_ae_p95 = float(np.quantile(err_tr, 0.95))
ae_f2, thr_ae_f2, (ae_p, ae_r, ae_f1_at_f2) = best_thr_fbeta(err_va, yva_bin_aligned, beta=2.0)
print(f"[AE] best F2={ae_f2:.3f} @ thr={thr_ae_f2:.6f} | P={ae_p:.3f} R={ae_r:.3f} (F1={ae_f1_at_f2:.3f})")

yhat_ae_p95 = eval_with_thr(err_va, yva_bin_aligned, thr_ae_p95, "AE-LSTM @thr p95 (unsupervised)", "ae_thr_p95")
yhat_ae_f2  = eval_with_thr(err_va, yva_bin_aligned, thr_ae_f2,  "AE-LSTM @thr F2 (valid-supervised)", "ae_thr_f2")

# ============================================
# 5) Isolation Forest (fit on NORMAL)
# ============================================
scaler_if = StandardScaler()
Xtr_if = scaler_if.fit_transform(X_train_df[mask_tr_normal])
Xva_if = scaler_if.transform(X_valid_df)

iforest = IsolationForest(
    n_estimators=400,
    contamination='auto',
    random_state=RANDOM_STATE,
    n_jobs=-1
).fit(Xtr_if)

scores_if = -iforest.score_samples(Xva_if)   # higher = more anomalous
scores_if_tr = -iforest.score_samples(Xtr_if)

if_roc_auc = roc_auc_score(y_valid_bin, scores_if)
if_pr_auc  = average_precision_score(y_valid_bin, scores_if)
print(f"\n[IForest] ROC-AUC={if_roc_auc:.3f} | PR-AUC={if_pr_auc:.3f}")
plot_roc_pr(y_valid_bin, scores_if, "IForest", "iforest")

thr_if_p95 = float(np.quantile(scores_if_tr, 0.95))
if_f2, thr_if_f2, (if_p, if_r, if_f1_at_f2) = best_thr_fbeta(scores_if, y_valid_bin, beta=2.0)
print(f"[IF] best F2={if_f2:.3f} @ thr={thr_if_f2:.6f} | P={if_p:.3f} R={if_r:.3f} (F1={if_f1_at_f2:.3f})")

yhat_if_p95 = eval_with_thr(scores_if, y_valid_bin, thr_if_p95, "IForest @thr p95 (unsupervised)", "if_thr_p95", cmap="Oranges")
yhat_if_f2  = eval_with_thr(scores_if, y_valid_bin, thr_if_f2,  "IForest @thr F2 (valid-supervised)", "if_thr_f2", cmap="Oranges")

# ============================================
# 6) Ensemble (AE + IF) and alert policy
# ============================================
# Normalize to [0,1] using TRAIN stats
ae_lo, ae_hi = err_tr.min(), err_tr.max()
ae_norm = minmax_transform(err_va, ae_lo, ae_hi)

# Align IF scores to AE sequence count / alignment
idx_start = LOOKBACK - 1
idx_end   = len(scores_if) - HORIZON_SHIFT  # normally equals len(scores_if)
scores_if_aligned = scores_if[idx_start:idx_end]

assert len(scores_if_aligned) == len(ae_norm) == len(yva_bin_aligned), \
    f"Misaligned: IF={len(scores_if_aligned)}, AE={len(ae_norm)}, y={len(yva_bin_aligned)}"

# Normalize IF with TRAIN stats
if_lo, if_hi = scores_if_tr.min(), scores_if_tr.max()
if_norm_aligned = minmax_transform(scores_if_aligned, if_lo, if_hi)

ALPHA = 0.6  # weight for AE (sequential model)
ens_score = ALPHA * ae_norm + (1 - ALPHA) * if_norm_aligned

# Threshold-free metrics for ensemble
ens_roc_auc = roc_auc_score(yva_bin_aligned, ens_score)
ens_pr_auc  = average_precision_score(yva_bin_aligned, ens_score)
print(f"\n[Ensemble] ROC-AUC={ens_roc_auc:.3f} | PR-AUC={ens_pr_auc:.3f}")
plot_roc_pr(yva_bin_aligned, ens_score, "Ensemble", "ensemble")

# Optimal threshold by F2 on ensemble + evaluation
ens_f2, thr_ens_f2, (ens_p, ens_r, ens_f1) = best_thr_fbeta(ens_score, yva_bin_aligned, beta=2.0)
print(f"[ENSEMBLE] best F2={ens_f2:.3f} @ thr={thr_ens_f2:.6f} | P={ens_p:.3f} R={ens_r:.3f} (F1={ens_f1:.3f})")
yhat_ens_f2 = eval_with_thr(ens_score, yva_bin_aligned, thr_ens_f2, "ENSEMBLE @thr F2", "ens_thr_f2")

# Alert smoothing: 3 out of 5 consecutive
alert_smooth = smooth_alerts(yhat_ens_f2, k=3, m=5)
print("\n[Ensemble + smoothing 3/5]")
print(classification_report(yva_bin_aligned, alert_smooth, target_names=["NORMAL","NO-NORMAL"], digits=4, zero_division=0))
cm = confusion_matrix(yva_bin_aligned, alert_smooth, labels=[0,1])
ConfusionMatrixDisplay(cm, display_labels=["NORMAL","NO-NORMAL"]).plot(cmap="PuBu")
plt.title("ENSEMBLE + Smoothing (3/5)")
save_show(PLOTS_DIR / "ens_smoothing_cm.png")

# ============================================
# 7) Interpretability
# ============================================
def ae_feature_contribs(model, Xseq: np.ndarray, feature_names: List[str]) -> pd.DataFrame:
    """Mean squared error per feature (time-averaged) as contribution."""
    rec = model.predict(Xseq, verbose=0)
    per_feat_mse = ((Xseq - rec) ** 2).mean(axis=1)
    return pd.DataFrame(per_feat_mse, columns=feature_names)

df_contrib = ae_feature_contribs(ae, Xva_seq, X_cols)
anom_idx = np.where(alert_smooth==1)[0]
topk = 10
if len(anom_idx) > 0:
    mean_contrib_anom = df_contrib.iloc[anom_idx].mean().sort_values(ascending=False).head(topk)
    print("\n[AE] Top features explaining anomalies (avg):")
    print(mean_contrib_anom)
    plt.figure()
    mean_contrib_anom[::-1].plot(kind="barh")
    plt.title("AE – Top contributions in anomalies")
    save_show(PLOTS_DIR / "ae_top_contrib_anom.png")

    i = anom_idx[0]
    plt.figure()
    df_contrib.iloc[i].sort_values(ascending=False).head(topk)[::-1].plot(kind="barh")
    plt.title("AE – Top contributions (example alert)")
    save_show(PLOTS_DIR / "ae_top_contrib_example.png")
else:
    print("\n[AE] No smoothed alerts to analyze contributions.")

# Surrogate tree to explain ensemble+smoothing decisions
X_valid_aligned = X_valid_df.iloc[LOOKBACK-1:LOOKBACK-1+len(alert_smooth)]
sur = DecisionTreeClassifier(max_depth=3, random_state=RANDOM_STATE)
sur.fit(X_valid_aligned, alert_smooth)
sur_imp = pd.Series(sur.feature_importances_, index=X_cols).sort_values(ascending=False).head(12)
print("\n[Surrogate Tree] Top features separating NORMAL/ANOMALO (ensemble+smoothing):")
print(sur_imp)
plt.figure()
sur_imp[::-1].plot(kind="barh")
plt.title("Surrogate – Importances")
save_show(PLOTS_DIR / "surrogate_importances.png")

# ============================================
# 8) Save artifacts (models, scalers, columns, meta, eval)
# ============================================
ae.save(ARTIFACTS_DIR / "ae_lstm.keras")
joblib.dump(iforest, ARTIFACTS_DIR / "iforest.pkl")
pd.Series(X_cols).to_csv(ARTIFACTS_DIR / "feature_columns.csv", index=False)
joblib.dump(scaler_ae, ARTIFACTS_DIR / "scaler_ae.pkl")
joblib.dump(scaler_if, ARTIFACTS_DIR / "scaler_if.pkl")
joblib.dump(le, ARTIFACTS_DIR / "label_encoder.pkl")

eval_df = pd.DataFrame({
    "timestamp": va_idx,
    "y_true": yva_bin_aligned,
    "ae_score": ae_norm,
    "if_score": if_norm_aligned,
    "ens_score": ens_score,
    "ens_pred_f2": yhat_ens_f2,
    "ens_pred_smooth": alert_smooth,
}).set_index("timestamp")
eval_df.to_parquet(ARTIFACTS_DIR / "eval_valid_window.parquet")

meta = {
    "lookback": LOOKBACK,
    "horizon_shift": HORIZON_SHIFT,
    "classes": le.classes_.tolist(),
    "normal_id": int(normal_id),

    # AE thresholds and metrics
    "ae_thr_p95": float(thr_ae_p95),
    "ae_thr_f2": float(ae_f2 if ae_f2 is not None else 0.0),
    "ae_roc_auc": float(ae_roc_auc),
    "ae_pr_auc": float(ae_pr_auc),

    # IF thresholds and metrics
    "if_thr_p95": float(thr_if_p95),
    "if_thr_f2": float(if_f2 if if_f2 is not None else 0.0),
    "if_roc_auc": float(if_roc_auc),
    "if_pr_auc": float(if_pr_auc),

    # Ensemble metrics and params
    "alpha": float(ALPHA),
    "ens_thr_f2": float(thr_ens_f2 if thr_ens_f2 is not None else 0.0),
    "ens_roc_auc": float(ens_roc_auc),
    "ens_pr_auc": float(ens_pr_auc),

    # Score normalization (TRAIN stats)
    "ae_score_min": float(ae_lo),
    "ae_score_max": float(ae_hi),
    "if_score_min": float(if_lo),
    "if_score_max": float(if_hi),

    # Alert policy
    "smoothing_k": 3,
    "smoothing_m": 5,
}
with open(ARTIFACTS_DIR / "meta.json", "w", encoding="utf-8") as f:
    json.dump(meta, f, ensure_ascii=False, indent=2)

# Optional: also save a horizon-tagged copy for later comparisons
meta_tagged_path = RESULTS_DIR / f"meta_h{HORIZON_SHIFT}.json"
with open(meta_tagged_path, "w", encoding="utf-8") as f:
    json.dump(meta, f, ensure_ascii=False, indent=2)

print("\n Artefactos guardados en:", ARTIFACTS_DIR)
print("   - Modelos: ae_lstm.keras, iforest.pkl")
print("   - Scalers: scaler_ae.pkl, scaler_if.pkl")
print("   - Encoder: label_encoder.pkl")
print("   - Columns: feature_columns.csv")
print("   - Eval DF: eval_valid_window.parquet")
print("   - Meta   : meta.json y", meta_tagged_path.name)
print("   - Plots  :", PLOTS_DIR)
```

```python
# ============================================
# Modelado Anomalías (AE-LSTM + IForest) con horizonte configurable
# ============================================
# Uso:
#   - En notebook: pega en una celda y ejecuta.
#   - Como script: python train_anomalia.py
#
# Requisitos: deltalake, numpy, pandas, scikit-learn, tensorflow, joblib, matplotlib
# ============================================

import os, json
from pathlib import Path
from typing import List
from collections import Counter, deque

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from deltalake import DeltaTable
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    classification_report, confusion_matrix, ConfusionMatrixDisplay,
    roc_auc_score, average_precision_score, precision_recall_curve, roc_curve
)
from sklearn.ensemble import IsolationForest
from sklearn.tree import DecisionTreeClassifier

import tensorflow as tf
from tensorflow.keras import layers, models, callbacks
import joblib

# -------------------------
# Configuración principal
# -------------------------
RANDOM_STATE   = 42
LOOKBACK       = 24     # pasos por secuencia (24h)
HORIZON_SHIFT  = 12     # 0=presente, 6/12/24=predicción a futuro (h)
OPERATE_WITH_AE_ONLY = True   # Recomendado para H=12 (mejores métricas en tus pruebas)
ALPHA          = 0.9    # Peso para AE si usas ensamble (si OPERATE_WITH_AE_ONLY=True, se ignora)

np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)
plt.rcParams["figure.figsize"] = (6, 4)

# -------------------------
# Rutas (auto-descubre raíz del repo)
# -------------------------
def find_repo_root() -> Path:
    p = Path.cwd().resolve()
    for cand in (p, *p.parents):
        if (cand / "data").exists():
            return cand
    return p

BASE_DIR = find_repo_root()
RUTA_GOLD_COMPLETE = BASE_DIR / "data" / "capa_gold" / "features_transformador" / "features_complete"

ARTIFACTS_DIR = BASE_DIR / "data" / "capa_gold" / "artifacts_anomalia"
PLOTS_DIR     = ARTIFACTS_DIR / "plots"
RESULTS_DIR   = BASE_DIR / "data" / "capa_gold" / "artifacts_anomalia_results"
for d in (ARTIFACTS_DIR, PLOTS_DIR, RESULTS_DIR):
    d.mkdir(parents=True, exist_ok=True)

# Columnas a excluir (labels/texto/futuras)
LEAK_OR_TEXT_COLS = [
    "estado_futuro","falla_30d","rul_dias","severidad_futura",
    "dias_proximo_evento","proximidad_evento","riesgo_acumulativo",
    "estado_operacional","nivel_severidad","variables_anomalas","descripcion_anomalia"
]
TARGET = "estado_futuro"

# -------------------------
# Utils: plots, thresholds, smoothing, scaling
# -------------------------
def save_show(figpath: Path):
    plt.tight_layout()
    plt.savefig(figpath, dpi=160)
    plt.show()

def plot_roc_pr(y_true: np.ndarray, scores: np.ndarray, title_prefix: str, out_prefix: str):
    fpr, tpr, _ = roc_curve(y_true, scores)
    prec, rec, _ = precision_recall_curve(y_true, scores)
    roc_auc = roc_auc_score(y_true, scores)
    pr_auc  = average_precision_score(y_true, scores)

    plt.figure()
    plt.plot(fpr, tpr); plt.plot([0,1],[0,1],'--')
    plt.title(f"{title_prefix} ROC (AUC={roc_auc:.3f})")
    plt.xlabel("FPR"); plt.ylabel("TPR")
    save_show(PLOTS_DIR / f"{out_prefix}_roc.png")

    plt.figure()
    plt.plot(rec, prec)
    plt.title(f"{title_prefix} Precision-Recall (AP={pr_auc:.3f})")
    plt.xlabel("Recall"); plt.ylabel("Precision")
    save_show(PLOTS_DIR / f"{out_prefix}_pr.png")

def eval_with_thr(scores: np.ndarray, y_true: np.ndarray, thr: float, title: str, out_prefix: str, cmap="Blues"):
    yhat = (scores > thr).astype(int)  # 1 = anómalo
    print(f"\n{title} (thr={thr:.6f})")
    print(classification_report(y_true, yhat, target_names=["NORMAL","NO-NORMAL"], digits=4, zero_division=0))
    cm = confusion_matrix(y_true, yhat, labels=[0,1])
    ConfusionMatrixDisplay(cm, display_labels=["NORMAL","NO-NORMAL"]).plot(cmap=cmap)
    plt.title(title)
    save_show(PLOTS_DIR / f"{out_prefix}_cm.png")
    return yhat

def best_thr_fbeta(scores: np.ndarray, y_true: np.ndarray, beta: float = 2.0, grid: int = 400):
    lo, hi = scores.min(), scores.max()
    ths = np.linspace(lo, hi, grid)
    best = (0.0, None, (0,0,0))  # (fbeta, thr, (precision, recall, f1))
    b2 = beta * beta
    for th in ths:
        yhat = (scores > th).astype(int)
        tp = ((y_true==1) & (yhat==1)).sum()
        fp = ((y_true==0) & (yhat==1)).sum()
        fn = ((y_true==1) & (yhat==0)).sum()
        prec = tp / (tp + fp) if (tp+fp)>0 else 0.0
        rec  = tp / (tp + fn) if (tp+fn)>0 else 0.0
        denom = b2*prec + rec
        fbeta = ((1+b2)*prec*rec/denom) if denom>0 else 0.0
        if fbeta > best[0]:
            f1 = (2*prec*rec)/(prec+rec) if (prec+rec)>0 else 0.0
            best = (fbeta, th, (prec, rec, f1))
    return best

def smooth_alerts(binary_seq: np.ndarray, k: int = 3, m: int = 5) -> np.ndarray:
    out = np.zeros_like(binary_seq)
    win = deque(maxlen=m)
    count = 0
    for i, v in enumerate(binary_seq):
        if len(win) == m:
            count -= win[0]
        win.append(v)
        count += v
        out[i] = 1 if count >= k else 0
    return out

def minmax_transform(x: np.ndarray, lo: float, hi: float):
    eps = 1e-12
    return (x - lo) / max(hi - lo, eps)

# -------------------------
# 1) Carga Gold y preparación
# -------------------------
dt = DeltaTable(str(RUTA_GOLD_COMPLETE))
df = dt.to_pandas()
df["timestamp"] = pd.to_datetime(df["timestamp"], utc=True, errors="coerce")
df = df.sort_values("timestamp").set_index("timestamp")

numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()
X_cols = [c for c in numeric_cols if c not in LEAK_OR_TEXT_COLS]

# descarta constantes y columnas con >30% NaN (para definir X_cols)
const_cols = [c for c in X_cols if df[c].nunique() <= 1]
if const_cols:
    print("Drop constantes:", const_cols)
    X_cols = [c for c in X_cols if c not in const_cols]

nan_ratio = df[X_cols].isna().mean()
high_nan = nan_ratio[nan_ratio > 0.30].index.tolist()
if high_nan:
    print("Drop >30% NaN:", high_nan)
    X_cols = [c for c in X_cols if c not in high_nan]

X_full = df[X_cols].replace([np.inf, -np.inf], np.nan)
y_raw = df["estado_futuro"].astype(str)

# Label encoder (mapea etiquetas a enteros manteniendo clases)
le = LabelEncoder()
y_enc = le.fit_transform(y_raw)
print("Clases:", dict(zip(le.classes_, range(len(le.classes_)))))

# -------------------------
# 2) Split temporal 80/20 sin fuga
# -------------------------
n = len(X_full); cut = int(n*0.80)
train_index = X_full.index[:cut]
valid_index = X_full.index[cut:]

# Imputación con medianas SOLO del TRAIN
train_medians = X_full.loc[train_index].median(numeric_only=True)
X_train_df = X_full.loc[train_index].fillna(train_medians)
X_valid_df = X_full.loc[valid_index].fillna(train_medians)

y_train_vec = le.transform(df.loc[train_index, "estado_futuro"].astype(str))
y_valid_vec = le.transform(df.loc[valid_index, "estado_futuro"].astype(str))

# Clase NORMAL
try:
    normal_id = int(np.where(le.classes_ == "NORMAL")[0][0])
except Exception:
    normal_id = Counter(y_train_vec).most_common(1)[0][0]

# binario (solo valid) 1=NO-NORMAL
y_valid_bin = (y_valid_vec != normal_id).astype(int)
print("\n== SPLIT ==")
print("Train:", X_train_df.shape, " Valid:", X_valid_df.shape)
print("Train range:", train_index.min(), "→", train_index.max())
print("Valid range:", valid_index.min(), "→", valid_index.max())
print("Distribución VALID (bin):", Counter(y_valid_bin))

# -------------------------
# 3) Escalado (fit en NORMAL) + secuencias
# -------------------------
mask_tr_normal = (y_train_vec == normal_id)

scaler_ae = StandardScaler()
X_train_sc = pd.DataFrame(
    scaler_ae.fit_transform(X_train_df[mask_tr_normal]),
    index=X_train_df.index[mask_tr_normal],
    columns=X_cols
)
X_valid_sc = pd.DataFrame(
    scaler_ae.transform(X_valid_df),
    index=X_valid_df.index,
    columns=X_cols
)

def make_sequences(X: pd.DataFrame, lookback: int):
    Xv = X.values; N, F = Xv.shape
    seqs, idx = [], []
    for t in range(lookback-1, N-HORIZON_SHIFT):
        seqs.append(Xv[t-(lookback-1):t+1])
        idx.append(X.index[t])  # timestamp final de la ventana
    return np.array(seqs, dtype=np.float32), pd.Index(idx, name="timestamp")

Xtr_seq, tr_idx = make_sequences(X_train_sc, LOOKBACK)
Xva_seq, va_idx = make_sequences(X_valid_sc, LOOKBACK)

# Etiquetas alineadas a secuencias y horizonte
yva_full = (y_valid_vec != normal_id).astype(int)
yva_bin_aligned = yva_full[LOOKBACK-1:len(yva_full)-HORIZON_SHIFT]

print("\n== SEQ ==")
print("Xtr_seq:", Xtr_seq.shape, "| Xva_seq:", Xva_seq.shape)
print("y_valid_bin aligned:", yva_bin_aligned.shape)

# -------------------------
# 4) Autoencoder LSTM (entrena con NORMAL)
# -------------------------
def build_lstm_ae(steps, feats):
    inp = layers.Input(shape=(steps, feats))
    x = layers.Masking()(inp)
    x = layers.LSTM(128, return_sequences=True)(x)
    x = layers.Dropout(0.2)(x)
    x = layers.LSTM(64)(x)
    z = layers.Dense(32, activation="relu")(x)
    x = layers.RepeatVector(steps)(z)
    x = layers.LSTM(64, return_sequences=True)(x)
    x = layers.LSTM(128, return_sequences=True)(x)
    out = layers.TimeDistributed(layers.Dense(feats))(x)
    m = models.Model(inp, out)
    m.compile(optimizer=tf.keras.optimizers.Adam(1e-3), loss="mse")
    return m

ae = build_lstm_ae(LOOKBACK, len(X_cols))
es  = callbacks.EarlyStopping(monitor="val_loss", patience=6, restore_best_weights=True)
rlr = callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=3, min_lr=1e-5)

hist = ae.fit(
    Xtr_seq, Xtr_seq,
    validation_data=(Xva_seq, Xva_seq),
    epochs=60, batch_size=128,
    callbacks=[es, rlr], verbose=1
)

plt.figure()
plt.plot(hist.history["loss"], label="Train loss")
plt.plot(hist.history["val_loss"], label="Valid loss")
plt.title("AE-LSTM Training vs Validation Loss")
plt.xlabel("Epoch"); plt.ylabel("MSE"); plt.legend()
save_show(PLOTS_DIR / "ae_training_loss.png")

def recon_error(model, Xseq):
    rec = model.predict(Xseq, verbose=0)
    return np.mean((Xseq - rec)**2, axis=(1,2))

err_tr = recon_error(ae, Xtr_seq)   # para normalizar y p95
err_va = recon_error(ae, Xva_seq)

ae_roc_auc = roc_auc_score(yva_bin_aligned, err_va)
ae_pr_auc  = average_precision_score(yva_bin_aligned, err_va)
print(f"\n[AE-LSTM] ROC-AUC={ae_roc_auc:.3f} | PR-AUC={ae_pr_auc:.3f}")
plot_roc_pr(yva_bin_aligned, err_va, "AE-LSTM", "ae")

thr_ae_p95 = float(np.quantile(err_tr, 0.95))
ae_f2, thr_ae_f2, (ae_p, ae_r, ae_f1_at_f2) = best_thr_fbeta(err_va, yva_bin_aligned, beta=2.0)
print(f"[AE] best F2={ae_f2:.3f} @ thr={thr_ae_f2:.6f} | P={ae_p:.3f} R={ae_r:.3f} (F1={ae_f1_at_f2:.3f})")

_ = eval_with_thr(err_va, yva_bin_aligned, thr_ae_p95, "AE-LSTM @thr p95 (unsupervised)", "ae_thr_p95")
_ = eval_with_thr(err_va, yva_bin_aligned, thr_ae_f2,  "AE-LSTM @thr F2 (valid-supervised)", "ae_thr_f2")

# -------------------------
# 5) Isolation Forest (fit en NORMAL)
# -------------------------
scaler_if = StandardScaler()
Xtr_if = scaler_if.fit_transform(X_train_df[mask_tr_normal])
Xva_if = scaler_if.transform(X_valid_df)

iforest = IsolationForest(
    n_estimators=400,
    contamination='auto',
    random_state=RANDOM_STATE,
    n_jobs=-1
).fit(Xtr_if)

scores_if    = -iforest.score_samples(Xva_if)   # mayor = más anómalo
scores_if_tr = -iforest.score_samples(Xtr_if)

if_roc_auc = roc_auc_score(y_valid_bin, scores_if)
if_pr_auc  = average_precision_score(y_valid_bin, scores_if)
print(f"\n[IForest] ROC-AUC={if_roc_auc:.3f} | PR-AUC={if_pr_auc:.3f}")
plot_roc_pr(y_valid_bin, scores_if, "IForest", "iforest")

thr_if_p95 = float(np.quantile(scores_if_tr, 0.95))
if_f2, thr_if_f2, (if_p, if_r, if_f1_at_f2) = best_thr_fbeta(scores_if, y_valid_bin, beta=2.0)
_ = eval_with_thr(scores_if, y_valid_bin, thr_if_p95, "IForest @thr p95 (unsupervised)", "if_thr_p95", cmap="Oranges")
_ = eval_with_thr(scores_if, y_valid_bin, thr_if_f2,  "IForest @thr F2 (valid-supervised)", "if_thr_f2", cmap="Oranges")

# -------------------------
# 6) Ensamble (AE + IF) y política operativa
# -------------------------
# Normalización [0,1] con stats de TRAIN
ae_lo, ae_hi = err_tr.min(), err_tr.max()
ae_norm = minmax_transform(err_va, ae_lo, ae_hi)

idx_start = LOOKBACK - 1
idx_end   = len(scores_if) - HORIZON_SHIFT
scores_if_aligned = scores_if[idx_start:idx_end]

assert len(scores_if_aligned) == len(ae_norm) == len(yva_bin_aligned), \
    f"Desalineado: IF={len(scores_if_aligned)}, AE={len(ae_norm)}, y={len(yva_bin_aligned)}"

if_lo, if_hi = scores_if_tr.min(), scores_if_tr.max()
if_norm_aligned = minmax_transform(scores_if_aligned, if_lo, if_hi)

ens_score = ALPHA * ae_norm + (1 - ALPHA) * if_norm_aligned
ens_roc_auc = roc_auc_score(yva_bin_aligned, ens_score)
ens_pr_auc  = average_precision_score(yva_bin_aligned, ens_score)
print(f"\n[Ensemble] ROC-AUC={ens_roc_auc:.3f} | PR-AUC={ens_pr_auc:.3f}")
plot_roc_pr(yva_bin_aligned, ens_score, "Ensemble", "ensemble")

ens_f2, thr_ens_f2, (ens_p, ens_r, ens_f1) = best_thr_fbeta(ens_score, yva_bin_aligned, beta=2.0)
_ = eval_with_thr(ens_score, yva_bin_aligned, thr_ens_f2, "ENSEMBLE @thr F2", "ens_thr_f2")

# --- Decisión operativa (recomendada AE-only para H=12) ---
if OPERATE_WITH_AE_ONLY:
    operate_score = ae_norm
    # Calcular threshold óptimo en la escala normalizada [0,1]
    ae_f2_norm, thr_ae_f2_norm, _ = best_thr_fbeta(operate_score, yva_bin_aligned, beta=2.0)
    operate_thr = float(thr_ae_f2_norm)
else:
    operate_score = ens_score
    _, operate_thr, _ = best_thr_fbeta(operate_score, yva_bin_aligned, beta=2.0)

yhat_operate = (operate_score > operate_thr).astype(int)

k, m = 3, 5  # smoothing por defecto
alert_operate = smooth_alerts(yhat_operate, k=k, m=m)
print("\n[Operative decision]")
print(classification_report(yva_bin_aligned, alert_operate, target_names=["NORMAL","NO-NORMAL"], digits=4, zero_division=0))
cm = confusion_matrix(yva_bin_aligned, alert_operate, labels=[0,1])
ConfusionMatrixDisplay(cm, display_labels=["NORMAL","NO-NORMAL"]).plot(cmap="PuBu")
plt.title("Operative decision + Smoothing")
save_show(PLOTS_DIR / "operative_smoothing_cm.png")

# -------------------------
# 7) Interpretabilidad (AE contribs + Surrogate tree)
# -------------------------
def ae_feature_contribs(model, Xseq: np.ndarray, feature_names: List[str]) -> pd.DataFrame:
    rec = model.predict(Xseq, verbose=0)
    per_feat_mse = ((Xseq - rec) ** 2).mean(axis=1)  # promedio temporal
    return pd.DataFrame(per_feat_mse, columns=feature_names)

df_contrib = ae_feature_contribs(ae, Xva_seq, X_cols)
anom_idx = np.where(alert_operate==1)[0]
topk = 10
if len(anom_idx) > 0:
    mean_contrib_anom = df_contrib.iloc[anom_idx].mean().sort_values(ascending=False).head(topk)
    print("\n[AE] Top features explicando anomalías (promedio):")
    print(mean_contrib_anom)
    plt.figure()
    mean_contrib_anom[::-1].plot(kind="barh")
    plt.title("AE – Top contribuciones en anomalías")
    save_show(PLOTS_DIR / "ae_top_contrib_anom.png")

    i = anom_idx[0]
    plt.figure()
    df_contrib.iloc[i].sort_values(ascending=False).head(topk)[::-1].plot(kind="barh")
    plt.title("AE – Top contribuciones (ejemplo alerta)")
    save_show(PLOTS_DIR / "ae_top_contrib_example.png")
else:
    print("\n[AE] No hay alertas operativas para analizar contribuciones.")

X_valid_aligned = X_valid_df.iloc[LOOKBACK-1:LOOKBACK-1+len(alert_operate)]
sur = DecisionTreeClassifier(max_depth=3, random_state=RANDOM_STATE)
sur.fit(X_valid_aligned, alert_operate)
sur_imp = pd.Series(sur.feature_importances_, index=X_cols).sort_values(ascending=False).head(12)
print("\n[Surrogate Tree] Top features que separan NORMAL/ANÓMALO (operativo):")
print(sur_imp)
plt.figure()
sur_imp[::-1].plot(kind="barh")
plt.title("Surrogate – Importancias (operativo)")
save_show(PLOTS_DIR / "surrogate_importances.png")

# -------------------------
# 8) Guardado de artefactos y meta
# -------------------------
ae.save(ARTIFACTS_DIR / "ae_lstm.keras")
joblib.dump(iforest, ARTIFACTS_DIR / "iforest.pkl")
pd.Series(X_cols).to_csv(ARTIFACTS_DIR / "feature_columns.csv", index=False)
joblib.dump(scaler_ae, ARTIFACTS_DIR / "scaler_ae.pkl")
joblib.dump(scaler_if, ARTIFACTS_DIR / "scaler_if.pkl")
joblib.dump(le, ARTIFACTS_DIR / "label_encoder.pkl")

# DF de evaluación con timestamp alineado (útil para auditorías/dashboards)
eval_df = pd.DataFrame({
    "timestamp": va_idx,
    "y_true": yva_bin_aligned,
    "ae_score": ae_norm,
    "if_score": if_norm_aligned,
    "operate_score": operate_score,
    "operate_pred_raw": yhat_operate,
    "operate_pred_smooth": alert_operate,
}).set_index("timestamp")
eval_df.to_parquet(ARTIFACTS_DIR / "eval_valid_window.parquet")

meta = {
    "lookback": LOOKBACK,
    "horizon_shift": HORIZON_SHIFT,
    "classes": le.classes_.tolist(),
    "normal_id": int(normal_id),

    # AE
    "ae_thr_p95": float(thr_ae_p95),
    "ae_thr_f2": float(ae_f2 if ae_f2 is not None else 0.0),
    "ae_roc_auc": float(ae_roc_auc),
    "ae_pr_auc": float(ae_pr_auc),

    # IF
    "if_thr_p95": float(thr_if_p95),
    "if_thr_f2": float(if_f2 if if_f2 is not None else 0.0),
    "if_roc_auc": float(if_roc_auc),
    "if_pr_auc": float(if_pr_auc),

    # Ensamble
    "alpha": float(ALPHA),
    "ens_thr_f2": float(ens_f2 if ens_f2 is not None else 0.0),
    "ens_roc_auc": float(roc_auc_score(yva_bin_aligned, ens_score)),
    "ens_pr_auc": float(average_precision_score(yva_bin_aligned, ens_score)),

    # Normalización (TRAIN)
    "ae_score_min": float(err_tr.min()),
    "ae_score_max": float(err_tr.max()),
    "if_score_min": float(scores_if_tr.min()),
    "if_score_max": float(scores_if_tr.max()),

    # Operación
    "operate_with_ae_only": bool(OPERATE_WITH_AE_ONLY),
    "operate_thr": float(operate_thr),
    "smoothing_k": int(k),
    "smoothing_m": int(m),
}
with open(ARTIFACTS_DIR / "meta.json", "w", encoding="utf-8") as f:
    json.dump(meta, f, ensure_ascii=False, indent=2)

# Copia etiquetada por horizonte (para comparativas h=0,6,12,24)
meta_tagged_path = RESULTS_DIR / f"meta_h{HORIZON_SHIFT}.json"
with open(meta_tagged_path, "w", encoding="utf-8") as f:
    json.dump(meta, f, ensure_ascii=False, indent=2)

print("\n Artefactos guardados en:", ARTIFACTS_DIR)
print("   - Modelos: ae_lstm.keras, iforest.pkl")
print("   - Scalers: scaler_ae.pkl, scaler_if.pkl")
print("   - Encoder: label_encoder.pkl")
print("   - Columns: feature_columns.csv")
print("   - Eval DF: eval_valid_window.parquet")
print("   - Meta   : meta.json y", meta_tagged_path.name)
print("   - Plots  :", PLOTS_DIR)
```
